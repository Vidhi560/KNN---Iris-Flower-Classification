# Train test split

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop(["target","flower_name"],axis ="columns")
y = df.target
# deleting data of specific column = for result 100%
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
#train_test_split=split in x & y
#random state=42 = randomly choose state of flower(42=gap)of 20% data 
#test_size=0.2 = 20%
#why we do = data will not biased
```

```python
len(x_train)
#total data 150 = train data 120 /80% 
```

```python
len(x_test)
#total data 150 = test data 30 / 20%
```

#create knn

```python
from sklearn.neighbors import KNeighborsClassifier
KNeighborsClassifier(n_neighbors =5)
#test in between near by 5 
```

```python
knn = KNeighborsClassifier(n_neighbors =5)
# train the data
```

```python
knn.fit(x_train, y_train)
#model ready by trained data
```

```python
knn.score(x_test,y_test)
#1.0 means 100% perfactly trained model 
```

```python
from sklearn.metrics import confusion_matrix
y_pred = knn.predict(x_test)
cm = confusion_matrix(y_test,y_pred)
```

```python
cm
```

```python
%matplotlib inline 
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize =(10,5))
sns.heatmap(cm,annot = True) # cm = array, (annot = true) = array in this
plt.xlabel("predicted")
plt.ylabel("Truth")
```
